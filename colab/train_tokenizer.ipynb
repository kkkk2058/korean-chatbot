```python
import os
import shutil

# 1. 코랩 컴퓨터 안에 data/ 폴더가 안전하게 있는지 확인하고 없으면 만듭니다.
os.makedirs("data", exist_ok=True)

# 2. 구글 드라이브 경로와 코랩 내부 경로 정의
drive_data_path = "/content/drive/MyDrive/korean-chatbot/train.txt"
colab_data_path = "data/train.txt"

# 3. 파일 복사 실행
if os.path.exists(drive_data_path):
    shutil.copy(drive_data_path, colab_data_path)
    print("🎉 구글 드라이브에서 train.txt 데이터를 성공적으로 가져왔습니다!")
else:
    print("❌ 구글 드라이브에 train.txt 파일이 없습니다. 1번 노트북을 먼저 실행해주세요.")
```

```python
from tokenizer import KoreanTokenizer

kt = KoreanTokenizer()
kt.train(files=["data/train.txt"], vocab_size=10000)
```

```python
import os
os.makedirs("tokenizer", exist_ok=True)
kt.save("tokenizer/tokenizer.json")
print(f"vocab size: {kt.vocab_size}")
```

```python
print(kt.encode("안녕하세요 저는 한국어 챗봇입니다"))
print(kt.decode(kt.encode("안녕하세요 저는 한국어 챗봇입니다")))
```
