```python
from src.tokenizer import BPETokenizer

tok = BPETokenizer()
tok.train(["data/namuwiki.txt"], vocab_size=8000)
```

```python
drive_dir = "/content/drive/MyDrive/korean-chatbot/models"
if os.path.exists(drive_dir):
    shutil.rmtree(drive_dir) # 기존 구버전 폴더가 있다면 삭제 후 최신화

# 파일 하나가 아닌 복사본 폴더 전체를 복사하여 merges.txt 등이 누락되는 것을 방지
shutil.copytree("models", drive_dir)
print("✅ 구글 드라이브에 토크나이저 모델 폴더 백업 완료!")
```
